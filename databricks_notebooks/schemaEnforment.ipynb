```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,LongType
```

```python
customer_schema=StructType([(StructField('customer_id',IntegerType(),True)),\
                            (StructField('customer_name',StringType(),False)),\
                            (StructField('city',StringType(),False)),\
                            (StructField('customer_email',StringType(),False)),\
                            (StructField('phone',LongType(),False))])
```

```python
customer_df=spark.read.option('header','true')\
            .schema(customer_schema)\
        .csv('/Volumes/workspace/default/test')
customer_df.show()
```

```python
customer_df.printSchema()
```

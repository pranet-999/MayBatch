```python
# I want to rank a customer based on their purchase within each city
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,rank,dense_rank
window_spec=Window.partitionBy('city').orderBy(col('total_revenue').desc())
ranking_df=total_revenue_df.withColumn('ranking',row_number().over(window_spec))
display(ranking_df)
```
